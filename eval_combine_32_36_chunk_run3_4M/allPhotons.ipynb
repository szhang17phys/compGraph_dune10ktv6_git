```python
# OpChannel Histogram Plotter
import matplotlib.pyplot as plt

# Step 1: Read the data file
bin_numbers = []
entry_counts = []

with open("./data/OpChannel_bin_counts.txt", "r") as file:
    for line in file:
        ch, count = map(int, line.strip().split())
        bin_numbers.append(ch)
        entry_counts.append(count / 2.0)  # average of 2 events


# Step 2: Plot
plt.figure(figsize=(18, 4), dpi=150) # default (12, 4)---
plt.bar(bin_numbers, entry_counts, width=1.0, color='red')

# Step 3: Labels and formatting
plt.xlabel("OpChannel")
plt.ylabel("Entries")
plt.xlim(0, 3750)
plt.ylim(0.9, 4000)
#plt.yscale('log')
plt.title("Entries per OpChannel")
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# Step 4: Show
plt.show()
```
